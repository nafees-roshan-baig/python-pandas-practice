```python
# 59. What is the purpose of np.arange(start, stop, step)? Explain each argument.

# np.arange() is used to generate a sequence of numbers with a fixed step size.
# start – Starting value (included).
# stop – Ending value (excluded).
# step – Difference between consecutive numbers.


# 60. What is the difference between np.zeros() and np.full()?

# np.zeros() creates an array filled with zeros.
# np.full() creates an array filled with a specified value.


# 61. What arguments are used in np.reshape(array, newshape)?

# It uses two arguments:
# array – The original array.
# newshape – The required shape of the array, such as rows and columns.


# 62. What is the difference between np.random.rand() and np.random.randn()?

# np.random.rand() generates random decimal values between 0 and 1.
# np.random.randn() generates random numbers following a standard normal distribution with mean 0 and standard deviation 1.


# 63. In np.random.randint(low, high, size), is the high value included?

# No, the high value is excluded. It generates random integers from low up to high-1.


# 64. What is the difference between np.max() and np.argmax()?

# np.max() returns the maximum value in an array.
# np.argmax() returns the index position of the maximum value.


# 65. What is the difference between np.min() and np.argmin()?

# np.min() returns the minimum value in an array.
# np.argmin() returns the index position of the minimum value.


# 66. What is the purpose of the axis argument in NumPy functions?

# The axis argument specifies the direction in which an operation is performed.
# axis=0 performs the operation column-wise.
# axis=1 performs the operation row-wise.


# 67. What is the difference between .loc[] and .iloc[] in Pandas?

# .loc[] is used to access data using index labels.
# .iloc[] is used to access data using integer positions.


# 68. What is the purpose of the index argument in pd.Series()?

# The index argument is used to assign custom labels to the elements of a Pandas Series instead of default numerical labels.


# 69. What do .head(n) and .tail(n) do?

# .head(n) displays the first n rows or elements.
# .tail(n) displays the last n rows or elements.


# 70. What is the difference between .shape, .size, and .ndim?

# .shape returns the number of rows and columns.
# .size returns the total number of elements.
# .ndim returns the number of dimensions.
```
