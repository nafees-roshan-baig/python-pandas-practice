```python
import pandas as pd
import numpy as np
data = np.array(['a','b','c','d'])
s = pd.Series(data)
print(s)
```

```python
data = np.array(['a','b','c','d'])
s = pd.Series(data,index=[100,101,102,103])
s 
```

```python
# create a dict in series
data = {'a':0,'b':1,'c':3}
s = pd.Series(data)
print(s)
```

```python
# 
data = {'a':0,'b':1,'c':2}
s = pd.Series(data,index=['b','c','d','a'],dtype = int)
print(s)
```

```python
#accessing datas from series
s = pd.Series([1,2,3,4,5])
print(s)
print(s[1])
```

```python
print(s[1:5])
```

```python
s = pd.Series([1,2,3,4,5],index = ['a','b','c','d','e'])
print(s['b':'d'])
print(s[['a','c','d']])
```

```python
#  creating an empty data frames
df = pd.DataFrame()
print(df)
```

```python
#  creating a data Frame from a list
data = [1,2,3,4,5]
df = pd.DataFrame(data)
print(df)
```

```python
# dataframe from dict
data = {'Name':['A','B','C','D'],'Age':[28,34,29,42]}
df = pd.DataFrame(data)
print(df)
```

```python
# DCreate a data from a dict of series
data = {'one':pd.Series([1,2,3],index=['a','b','c']),
        'two':pd.Series([1,2,3,4],index=['a','b','c','d'])}
df = pd.DataFrame(data)
print(df)
# select a single column
print(df['one'])
print(df['two'])
print(df[['one','two']])
```

```python
# adding new column
df['three'] = pd.Series([10,20,30],index=['a','b','c'])
df
```
