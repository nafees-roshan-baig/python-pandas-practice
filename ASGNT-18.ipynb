```python
#   1. Create Series
# Q1: Create a Pandas Series from a list [10, 20, 30, 40].
import pandas as pd
a=pd.Series([10,20,30,40])
print(a)
# Q2: Create a Series with custom indices: values [100, 200, 300] and index ['a', 'b', 'c'].
b=pd.Series([1,2,3],index=['a','b','c'])
print(b)
```

```python
#   2. Access Elements in Series
# Q1: Access the second element in the Series s = pd.Series([5, 10, 15, 20]).
s = pd.Series([5, 10, 15, 20])
print(s[1])
# Q2: Access elements from index 1 to 3 (inclusive).
print(s[1:3])
```

```python
#  3. Retrieve Elements from Series
# Q1: Retrieve element at index 'b' from a labeled Series s = pd.Series([1, 2, 3], index=['a', 'b', 'c']).
s = pd.Series([1, 2, 3], index=['a', 'b', 'c'])
print(s['b'])
# Q2: Retrieve all values greater than 1 from Series.
print(s[s>1])
```

```python
#   4. Create DataFrame
# Q1: Create a DataFrame from a dictionary: {'Name': ['A', 'B'], 'Age': [25, 30]}.
import pandas as pd
import numpy as np
df = pd.DataFrame({'name':('A','B'),'age':(25,30)})
df
# Q2: Create a DataFrame using a list of lists: [[1, 'Tom'], [2, 'Jerry']] with column names ['ID', 'Name'].
df= pd.DataFrame([[1, 'Tom'], [2, 'Jerry']],columns=['ID','Name'])
df
```

```python
# Q1: Create a DataFrame from a dictionary: {'Name': ['A', 'B'], 'Age': [25, 30]}.
df = pd.DataFrame({'Name': ['A', 'B'], 'Age': [25, 30]})
print(df)
# Q2: Select the first row using .iloc.
print(df.iloc[0])
```

```python
# # Basic Functionality in a Pandas Series
# Assume a series:
# s = pd.Series([10, 20, 30, 40], index=['a', 'b', 'c', 'd'])
s = pd.Series([10,20,30,40],index=['a','b','c','d'])
s
# Q1: Print the axes (row labels) of the Series. 
print(s.axes)
# Q1: Check the data type of the Series.
print(s.dtype)
# Q1: Check whether the Series is empty.
print(s.empty)
# Q1: Print number of dimensions of the Series.
print(s.ndim)
# Q1: Print the size (number of elements) of the Series.
print(s.size)
# Q1: Return all the values in the Series as NumPy array.
print(s.values)
# Q1: Display the first 2 values.
print(s.values[0:2])
# Q2: Display the last 2 values.
print(s.values[2:])
```

```python
# Basic Functionality in a Pandas DataFrame
# Assume a DataFrame:
# df = pd.DataFrame({
#     'Name': ['Alice', 'Bob', 'Charlie'],
#     'Age': [25, 30, 35]
# })

df = pd.DataFrame({'Name':['Alice', 'Bob', 'Charlie'],
                   'Age':[25,30,35]})
print(df)
# Q1: Transpose the DataFrame.
print(df.transpose())
# Q1: Print axes of DataFrame.
print(df.axes)
# Q1: Check the data types of all columns.
print(df.dtypes)
# Q1: Check if the DataFrame is empty.
print(df.empty)
# Q1: Print number of dimensions.
print(df.ndim)
# Q1: Print number of rows and columns.
print(df.shape)
# Q1: Print total number of elements in DataFrame.
print("no of ele",df.size)
# Q1: Display DataFrame values as a NumPy array.
print(df.values)
# Q1: Show first 2 rows.
print(df.iloc[0:2])
print(df.head(2))
# Q2: Show last 2 rows.
print(df.iloc[1:])
print(df.tail(2))
```
